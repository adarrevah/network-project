```python
import pandas as pd
from scapy.all import *
```

```python
df=pd.read_csv("group01_http_input.csv")
```

```python
df
```

```python
msg = df.iloc[0]['message']
```

```python
packet = IP(dst="127.0.0.1") / TCP(dport=80) / msg
```

```python
packet.show()
```

```python
print("start to send the packets")
for index, row in df.iterrows():
    p = IP(dst="127.0.0.1") / TCP(dport=80) / str(row['message'])
    send(p, verbose=False)
print("all the packets sent")
```
